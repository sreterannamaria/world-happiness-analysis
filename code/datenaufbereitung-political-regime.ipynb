# Datenaufbereitung: politisches System

Liest die Klassifizierung der politischen Systeme ein (0 = geschlossene Autokratie bis 3 = liberale Demokratie) und speichert sie in der Tabelle `metric`.

```python
import pandas as pd

from bina_db import connect, CountryLookup

pd.set_option('display.max_rows', 250)
```

```python
pol_reg = pd.read_csv("../data/additional/political-regime.csv")
```

```python
pol_reg
```

```python
metric_definition_id = 21  # Political System

maria_db_connection = connect()
country_lookup = CountryLookup(maria_db_connection)

insert_statement = "insert into metric(value, year, country, metric_definition) VALUES (?, ?, ?, ?)"

cursor = maria_db_connection.cursor()

for index, row in pol_reg.iterrows():
    country_name = row["Entity"]
    year = row["Year"]
    value = row["Political regime"]

    country_id = country_lookup.lookup(country_name)

    if country_id is None:
        print("Could not resolve country: " + country_name)
        continue

    cursor.execute(insert_statement, (value, year, country_id, metric_definition_id))

    maria_db_connection.commit()

maria_db_connection.close()
```
